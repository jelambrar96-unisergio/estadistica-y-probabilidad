# Distribución Chi-cuadrada ($\chi^2$)

La distribución Chi-cuadrada surge de la suma de cuadrados de variables normales estándar independientes. Es fundamental en la inferencia estadística, especialmente para pruebas de bondad de ajuste y pruebas de independencia.

### Definición matemática

Si $Z_1, Z_2, \dots, Z_k$ son variables aleatorias normales estándar independientes, entonces la variable:

$$X = \sum_{i=1}^k Z_i^2$$

sigue una distribución Chi-cuadrada con $k$ grados de libertad ($df=k$). Su PDF para $x > 0$ es:

$$f(x) = \frac{1}{2^{k/2} \Gamma(k/2)} x^{(k/2)-1} e^{-x/2}$$

### Características

- **Media**: $E[X] = k$
- **Varianza**: $Var[X] = 2k$
- **Asimetría**: Siempre es asimétrica a la derecha, aunque se vuelve más simétrica a medida que $k$ aumenta.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

## Ejercicio: Graficar múltiples distribuciones con diferentes parámetros

x = np.linspace(0, 30, 1000)
grados_libertad = [2, 5, 10, 15]
colores = ['blue', 'green', 'red', 'orange']

plt.figure(figsize=(10, 6))

for df, color in zip(grados_libertad, colores):
    pdf = chi2.pdf(x, df)
    plt.plot(x, pdf, label=f'Chi-cuadrada(df={df})', color=color, lw=2)
    plt.fill_between(x, pdf, color=color, alpha=0.1)

plt.title('Comparación de Distribuciones Chi-cuadrada')
plt.xlabel('x')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
