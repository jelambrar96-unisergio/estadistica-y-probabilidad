# Distribución F (Fisher-Snedecor)

La distribución F es la distribución del cociente de dos variables Chi-cuadrada independientes, cada una dividida por sus grados de libertad. Se utiliza principalmente en el **Análisis de Varianza (ANOVA)** para comparar si las varianzas de dos poblaciones son iguales o si las medias de varias poblaciones son diferentes.

### Definición matemática

Si $U$ y $V$ son variables aleatorias Chi-cuadrada independientes con $d_1$ y $d_2$ grados de libertad respectivamente, entonces:

$$X = \frac{U/d_1}{V/d_2}$$

sigue una distribución F con $(d_1, d_2)$ grados de libertad.

### Características

- **Media**: $E[X] = \frac{d_2}{d_2-2}$ (para $d_2 > 2$)
- **Rango**: Solo definida para $x \ge 0$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

## Ejercicio: Graficar múltiples distribuciones con diferentes parámetros

x = np.linspace(0, 5, 1000)
parametros = [(1, 10), (5, 10), (10, 10), (100, 100)]
colores = ['blue', 'green', 'red', 'orange']

plt.figure(figsize=(10, 6))

for (dfn, dfd), color in zip(parametros, colores):
    pdf = f.pdf(x, dfn, dfd)
    plt.plot(x, pdf, label=f'F(dfn={dfn}, dfd={dfd})', color=color, lw=2)
    plt.fill_between(x, pdf, color=color, alpha=0.1)

plt.title('Comparación de Distribuciones F')
plt.xlabel('x')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
