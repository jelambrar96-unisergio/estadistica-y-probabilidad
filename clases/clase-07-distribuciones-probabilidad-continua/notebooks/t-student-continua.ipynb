# Distribución t de Student

La distribución t de Student es similar a la normal pero con colas más pesadas. Se usa cuando queremos estimar la media de una población distribuida normalmente en situaciones donde el tamaño de la muestra es pequeño ($n < 30$) y la desviación estándar de la población es desconocida.

### Definición matemática

Si $Z$ es una normal estándar y $V$ es una Chi-cuadrada con $k$ grados de libertad, y son independientes, entonces:

$$T = \frac{Z}{\sqrt{V/k}}$$

sigue una distribución t con $k$ grados de libertad. Su PDF es simétrica respecto a cero.

### Características

- **Media**: $E[X] = 0$ (para $k > 1$)
- **Varianza**: $Var[X] = \frac{k}{k-2}$ (para $k > 2$)
- **Simetría**: Campana simétrica centrada en 0. A medida que $k \to \infty$, la distribución t converge a la normal estándar.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm

## Ejercicio: Graficar múltiples distribuciones con diferentes parámetros

x = np.linspace(-5, 5, 1000)
grados_libertad = [1, 2, 5, 30]
colores = ['blue', 'green', 'red', 'orange']

plt.figure(figsize=(10, 6))

# Graficar Normal Estándar como referencia
plt.plot(x, norm.pdf(x), 'k--', label='Normal Estándar', alpha=0.5, lw=2)

for df, color in zip(grados_libertad, colores):
    pdf = t.pdf(x, df)
    plt.plot(x, pdf, label=f't-Student(df={df})', color=color, lw=2)

plt.title('Comparación de Distribuciones t de Student vs Normal')
plt.xlabel('x')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
