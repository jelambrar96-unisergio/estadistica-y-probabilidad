# Distribución Normal

La distribución normal, también conocida como distribución de Gauss o "campana de Gauss", es la distribución más importante en estadística. Muchos fenómenos naturales y sociales siguen esta distribución (estatura, errores de medición, puntajes de exámenes).

### Definición matemática

Una variable aleatoria continua $X$ sigue una distribución normal con media $\mu$ y varianza $\sigma^2$ si su PDF es:

$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2}, \quad -\infty < x < \infty$$

### Características

- **Media**: $E[X] = \mu$
- **Varianza**: $Var[X] = \sigma^2$
- **Simetría**: Es perfectamente simétrica respecto a $\mu$. La media, mediana y moda coinciden.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

## Ejercicio: Graficar múltiples distribuciones con diferentes parámetros

x = np.linspace(-10, 10, 1000)
parametros = [(0, 1), (0, 2), (2, 1)]
colores = ['blue', 'green', 'red']

plt.figure(figsize=(10, 6))

for (mu, sigma), color in zip(parametros, colores):
    pdf = norm.pdf(x, loc=mu, scale=sigma)
    plt.plot(x, pdf, label=f'Normal($\mu={mu}, \sigma={sigma}$)', color=color, lw=2)
    plt.fill_between(x, pdf, color=color, alpha=0.1)

plt.title('Comparación de Distribuciones Normales')
plt.xlabel('x')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
