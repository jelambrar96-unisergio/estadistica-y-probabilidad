# Distribución Uniforme Continua

La distribución uniforme continua es la más simple de las distribuciones continuas. Describe una situación en la que todos los intervalos de igual longitud en el rango de la variable tienen la misma probabilidad de ocurrir. Se suele usar cuando sabemos que algo ocurrirá dentro de un rango, pero no tenemos razones para creer que un valor es más probable que otro.

### Definición matemática

Una variable aleatoria $X$ tiene una distribución uniforme en el intervalo $[a, b]$ si su función de densidad es:

$$
f(x) = \begin{cases}
\frac{1}{b-a} & \text{si } a \le x \le b \\
0 & \text{en otro caso}
\end{cases}
$$

### Características

- **Media**: $\mu = \frac{a+b}{2}$
- **Varianza**: $\sigma^2 = \frac{(b-a)^2}{12}$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

## Ejercicio: Graficar múltiples distribuciones con diferentes parámetros

x = np.linspace(-5, 25, 1000)
parametros = [(0, 10), (5, 15), (0, 20)]
colores = ['blue', 'green', 'red']

plt.figure(figsize=(10, 6))

for (a, b), color in zip(parametros, colores):
    # loc = a, scale = b - a
    pdf = uniform.pdf(x, loc=a, scale=b-a)
    plt.plot(x, pdf, label=f'Uniforme({a}, {b})', color=color, lw=2)
    plt.fill_between(x, pdf, color=color, alpha=0.1)

plt.title('Comparación de Distribuciones Uniformes Continuas')
plt.xlabel('x')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
