## Visualización para distintas tasas

```python
import numpy as np
import matplotlib.pyplot as plt

lambdas = [1, 3, 6]
k_max = 15
x = np.arange(0, k_max + 1)

plt.figure(figsize=(10, 5))
for lam in lambdas:
    p = [poisson_pmf(k, lam) for k in x]
    plt.plot(x, p, marker='o', label=f'λ = {lam}')

plt.title('Distribución de Poisson para diferentes valores de λ')
plt.xlabel('k')
plt.ylabel('P(X = k)')
plt.legend()
plt.grid(True)
plt.show()
```
