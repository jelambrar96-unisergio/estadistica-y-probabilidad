## Visualización de la distribución

```python
import numpy as np
import matplotlib.pyplot as plt

r = 3
ps = [0.3, 0.5]
k_values = np.arange(r, r + 15)

plt.figure(figsize=(10, 5))
for p, color in zip(ps, ['r', 'b']):
    pmf_vals = [negative_binomial_pmf(r, p, k) for k in k_values]
    plt.stem(k_values, pmf_vals, label=f'p = {p}', linefmt=f'{color}-', markerfmt=f'{color}o', basefmt=' ')

plt.title('Distribución Binomial Negativa para diferentes valores de p')
plt.xlabel('k (ensayos hasta r éxitos)')
plt.ylabel('P(X = k)')
plt.legend()
plt.grid(True)
plt.show()
```
