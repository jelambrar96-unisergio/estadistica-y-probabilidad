```python
import numpy as np
import matplotlib.pyplot as plt

N, K, n = 25, 10, 5
x = np.arange(max(0, n - (N - K)), min(n, K) + 1)
prob = [hipergeometrica_pmf(N, K, n, k) for k in x]

plt.figure(figsize=(8, 4))
plt.bar(x, prob, color='salmon', edgecolor='black')
plt.title('Distribución Hipergeométrica')
plt.xlabel('k (éxitos en la muestra)')
plt.ylabel('P(X = k)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
