# Distribuciones probabilidad uniforme


### 1.1. Introducción

La distribución uniforme discreta describe un experimento en el cual todos los resultados tienen la misma probabilidad. Es útil para modelar situaciones en las que no hay preferencia entre las opciones.

### 1.2. Definición matemática

Si una variable aleatoria discreta X toma valores en el conjunto finito:

$(\{a, a+1, \dots, b\})$,

con $a$ y $b$ enteros y $a \le b$, entonces:

$$
P(X = x) = \begin{cases}
\frac{1}{b-a+1} & \text{si } x \in \{a, a+1, \dots, b\} \\
0 & \text{en otro caso}
\end{cases}
$$

### 1.3. Características

- Media:

$$\mu = \frac{a+b}{2}$$

- Varianza:
  
$$\sigma^2 = \frac{(b-a+1)^2-1}{12}$$

## 1. Graficando una distribucion de probabilidad uniforma discreta

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 1, 6
x = np.arange(a, b + 1)
p = np.ones_like(x, dtype=float) / len(x)

plt.figure(figsize=(8, 4))
plt.bar(x, p, color='skyblue', edgecolor='black')
plt.title('Distribución uniforme discreta: dado justo')
plt.xlabel('x')
plt.ylabel('P(X = x)')
plt.xticks(x)
plt.ylim(0, max(p) + 0.05)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## 3. Haciendo una muestra de una distribucion de probabilidad uniforma discreta

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 1, 6
muestra = np.random.choice(np.arange(a, b + 1), size=1000, replace=True)
conteos, _ = np.histogram(muestra, bins=np.arange(a, b + 2) - 0.5)

plt.figure(figsize=(8, 4))
plt.bar(valores, conteos / len(muestra), color='lightgreen', edgecolor='black')
plt.title('Frecuencia empírica de una muestra uniforme discreta')
plt.xlabel('x')
plt.ylabel('Frecuencia relativa')
plt.xticks(valores)
plt.ylim(0, 0.25)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
