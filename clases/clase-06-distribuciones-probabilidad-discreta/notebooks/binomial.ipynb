## Visualización de la distribución binomial

```python
import numpy as np
import matplotlib.pyplot as plt

n = 10
ps = [0.2, 0.4, 0.7]
ks = np.arange(0, n + 1)

plt.figure(figsize=(10, 5))
for p, color in zip(ps, ['r', 'b', 'g']):
    pmf_vals = [binomial_pmf(n, p, k) for k in ks]
    plt.stem(ks, pmf_vals, label=f'p = {p}', linefmt=f'{color}-', markerfmt=f'{color}o', basefmt=' ')

plt.title('Distribución Binomial para diferentes valores de p')
plt.xlabel('k (número de éxitos)')
plt.ylabel('P(X = k)')
plt.legend()
plt.grid(True)
plt.show()
```
